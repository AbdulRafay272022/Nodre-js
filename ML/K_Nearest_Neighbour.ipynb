KNN is a supervised machine learning algorithm that can be used to solve both classification and regression problems.

It is a non-parametric, lazy learning algorithm. Non-parametric means that it does not make any assumptions on the underlying data distribution. Lazy learning means that it does not require any training data points for model generation. All training data used in the testing phase. This makes training faster and testing phase slower and costlier.

KNN algorithm at the training phase just stores the dataset and when it gets new data, then it classifies that data into a category that is much similar to the new data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
df.info()
```

```python
x=df.drop('species',axis=1)
y=df['species']
```

```python
#load KNN Classifier
from sklearn.neighbors import KNeighborsClassifier

model=KNeighborsClassifier(n_neighbors=5)
model.fit(x,y)

model.predict([[6.4,3.2,4.5,1.5]])
```

```python
#abd train test split krkty lagao
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

model.fit(x_train,y_train)
y_pre=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report

print(confusion_matrix(y_test,y_pre))
print(classification_report(y_test,y_pre))
```

```python
sns.heatmap(confusion_matrix(y_test,y_pre),annot=True,fmt='d',cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.show()
```

```python
#Regresssion on tips data set
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import LabelEncoder
df_t=sns.load_dataset('tips')
df_t.info()
```

```python
LE=LabelEncoder()
for i in df_t.columns:
    if df_t[i].dtype=='category':
        df_t[i]=LE.fit_transform(df_t[i])
df_t        
```

```python
x=df_t.drop('tip',axis=1)
y=df_t['tip']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model_r=KNeighborsRegressor(n_neighbors=5,metric='minkowski',p=2)#ab yh minkowski ka euclidean use kare ga
model_r.fit(x_train,y_train)
y_pre=model_r.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error
print(r2_score(y_test,y_pre))
print(mean_squared_error(y_test,y_pre))
```
