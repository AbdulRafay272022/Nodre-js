```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.linear_model import LinearRegression
```

```python
df=sns.load_dataset('tips')
df
```

```python
#separating features and target
x=df[['total_bill','sex','time']]#regression ik numeric sy dosri numeric predict 
y=df['tip']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
#preprocessing train test split sy phale bhi hoskti bhad bhi
lb=LabelEncoder()
x_train['sex']=lb.fit_transform(x_train['sex'])
x_train['time']=lb.fit_transform(x_train['time'])
x_test['sex']=lb.fit_transform(x_test['sex'])
x_test['time']=lb.fit_transform(x_test['time'])
#scalar=StandardScaler()
#x_train=scalar.fit_transform(x_train)
#x_test=scalar.fit_transform(x_test)
#x_test
```

```python
#model bluid
model=LinearRegression()
model.fit(x_train,y_train)#sheek raha
y_pre=model.predict(x_test)#predict kr raha
```

```python
#evaluate model
from sklearn.metrics import mean_squared_error,r2_score#used for linear regression
print('MSE:',mean_squared_error(y_test,y_pre))
print('R2_Score:',r2_score(y_test,y_pre))
```

```python
model.predict([[100,1,1]])#ab ziada acha tw scaling krty hainmagar dhkety ky model kesa perform kr raha us phr
```
