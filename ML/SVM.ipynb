```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.svm import SVC,SVR
from sklearn.model_selection import train_test_split
from sklearn.metrics import  classification_report,confusion_matrix
from sklearn.preprocessing import LabelEncoder
```

```python
df=sns.load_dataset('iris')
```

```python
x=df.drop('species',axis=1)
y=df['species']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
model=SVC()
model.fit(x_train,y_train)
y_pre=model.predict(x_test)
```

```python
print(classification_report(y_test,y_pre))
print(confusion_matrix(y_test,y_pre))
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(confusion_matrix(y_test,y_pre),annot=True,fmt='d')
plt.xlabel('Predicted')
plt.ylabel('Truth')
plt.show()
```

```python
df1=sns.load_dataset('tips')
df1
```

```python
le=LabelEncoder()
df1['sex']=le.fit_transform(df1['sex'])
df1['smoker']=le.fit_transform(df1['smoker'])
df1['day']=le.fit_transform(df1['day'])
df1['time']=le.fit_transform(df1['time'])
```

```python
df1
```

```python
x=df1.drop('tip',axis=1)
y=df1['tip']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
model1=SVR()
model1.fit(x_train,y_train)
y_pre=model1.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
print(r2_score(y_test,y_pre))
print(mean_squared_error(y_test,y_pre))
print(mean_absolute_error(y_test,y_pre))
print(np.sqrt(mean_squared_error(y_test,y_pre)))
```
